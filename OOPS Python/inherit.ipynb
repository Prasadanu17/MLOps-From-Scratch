```python
# simple inheritance

# Base class
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print(f"{self.name} make a sound")

# Derived class 
class Dog(Animal):
    def speak(self):
        print(f"{self.name} barks")

# create an instance of Animal 
animal = Animal("Generic Animal")
animal.speak()

# create an instance of Dog
dog = Dog("Buddy")
dog.speak()
```
