```python
# simple inheritance

# Base class
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print(f"{self.name} make a sound")

# Derived class 
class Dog(Animal):
    def speak(self):
        print(f"{self.name} barks")

# create an instance of Animal 
animal = Animal("Generic Animal")
animal.speak()

# create an instance of Dog
dog = Dog("Buddy")
dog.speak()
```

```python
class College:
    def __init__(self):
        self.name = "ICFAI"
    def speak(self):
        print(f"{self.name} University")

class Location(College):
    def __init__(self, locat):
        self.name = "ICFAI"   # set it directly
        self.locat = locat
    def speak(self):
        print(f"Location of {self.name} University is {self.locat}")

loc = Location("Sichey")
loc.speak()   # Location of ICFAI University is Sichey
```

```python
# Super keyword

# Base class
class Animal:
    def __init__(self):
        self.name = "Buddy"

    def speak(self):
        print(f"{self.name} make a sound")

# Derived class 
class Dog(Animal):
    def __init__(self,breed):
        super().__init__()
        self.bread = breed

    def speak(self):
        super().speak()
        print(f"{self.name} barks. It is a {self.bread}.")

# create an instance of Dog
dog = Dog("Golden retriever")
dog.speak()
```

```python
# single or Basic Inheritance

# Base class
class parent:
    def __init__(self,name):
        self.name = name
    def greet(self):
        print(f"Hello, my name is {self.name}.")

# Derived class
class child(parent):

    def play(self):
        print(f"{self.name} is playing")

# create an instance of child
children = child("alice")
children.greet()
children.play()
```

```python
# Multi level inheritance 

# base class
class Grandparent:
    def __init__(self,name):
        self.name = name
    def tell_story(self):
        print(f"{self.name} tells a story")

# intermediate class 
class parent(Grandparent):
    def work(self):
        print(f"{self.name} is working")

# Derived class
class child(parent):
    def play(self):
        print(f"{self.name} is playing")

# create an instance of child 
child1=child("shyere")
child1.tell_story()
child1.work()
child1.play()
```

```python
# Hierarchical Inheritance 

# Base class 
class parent:
    def __init__(self, name):
        self.name=name

    def greet(self):
        print(f"Hello, my name is {self.name}.")

#  Derived class
class child1(parent):
     def play(self):
        print(f"{self.name} is playing.")

# Derived class 2 
class child2(parent):
    def study(self):
        print(f"{self.name} is studying.")

# create instance of child and child 2
child1 = child1("dave")
Child2 = child2("Eve")

child1.greet() 
child1.play()

Child2.greet()
Child2.study()
```

```python
# Multiple Inheritance

# common base class
class A:
    def __init__(self,name):
        self.name=name

    def greet(self):
        print(f"Hello from A, {self.name}.")

# Intermediate class 1
class B(A):
    def greet(self):
        print(f"hello from B, {self.name}.")
        super().greet()

# Intermediate class 2
class C(A):
    def greet(self):
        print(f"Hello from C, {self.name}.")
        super().greet()

# Derived class 
class D(B,C):
    def greet(self):
        print(f"Hello from D, {self.name}.")
        super().greet()

# create an instance D 
d= D("Frank")
d.greet()
```

```python
#  Hybrid Inheritance

# Base class 
class Animal:
    def __init__(self,name):
        self.name=name

    def sound(self):
        print(f"{self.name} makes a sound")

# Intermediate class 1 (Hierachical)
class Mammal(Animal):
    def feed(self):
        print(f"{self.name} is feeding milk.")

# Intermediate class 2 (Multiple)
class Bird(Animal):
    def fly(self):
        print(f"{self.name} is flying.")

# Derived class (Multiple Inheritance)
class Bat(Mammal,Bird):
    def __init__(self,name):
        Mammal.__init__(self,name) # Explicity calling the constructor

    def nocturnal(self):
        print(f"{self.name} is nocturnal.")

# create an instance of Bat
bat = Bat("Bruce")
bat.sound()
bat.feed()
bat.fly()
bat.nocturnal()
```
