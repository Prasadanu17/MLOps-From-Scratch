```python
class employee :
    # special methos/magic method//dunder method -constructor
    def __init__(self):
        print("Started executing attribute/data")
        self.id=33
        self.salary=50000
        self.designation ="SDE"
        print("attributes/data have been initialized")

    def travel(self,destination):
        print("This travel method was called manually")
        print(f"Employee is travelling to {destination}")
# create an object/instance of class
sam=employee()

# printing the attributes
# print(sam.id)

# calling the method 
sam.travel("chennai")

 
```
